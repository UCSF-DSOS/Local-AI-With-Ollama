```python
import ollama
```

```python
text = """
A federal judge issued a preliminary injunction preventing Philip Morris
from implementing portions of its Retail Leaders program after competitors
argued that the program would unfairly restrict cigarette advertising
and retail display space.
"""
```

```python
model = "strands-decider:2b"
```

```python
questions={
        "document_type": {
            "type": "choice",
            "instructions": "Classify this document into one category.",
            "criteria": {
                "legal": "Legal proceedings, court decisions, or litigation",
                "news": "News reporting or journalism",
                "advertising": "Advertising or promotional material",
                "corporate": "Internal corporate communication",
                "research": "Scientific or academic research",
                "other": "None of the above"
            }
        }
    }
```

```python
%%time
response = ollama.systemone(
    model=model,
    state=text,
    questions=questions
)
```

```python
print(response.answers["document_type"])
```
