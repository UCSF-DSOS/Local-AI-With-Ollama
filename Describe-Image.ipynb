```python
import ollama
```

```python
model = "qwen3-vl:2b-instruct"
```

```python
prompt = "Describe this image in detail."
```

```python
%%time
response = ollama.generate(
    model=model, 
    prompt=prompt,
    images=["./media/glnj0229.png"],
    options={
        "num_ctx": 8192,         # Expands the memory/context window (default is often 2048)
        "num_predict": 4096,     # Allows the model to output a much longer response
        "temperature": 0.7       # Encourages detailed, descriptive writing
    }
)
```

```python
print(response["response"])
```
