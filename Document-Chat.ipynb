```python
from pathlib import Path
import ollama
```

```python
document_path = Path("./media/pmcj0045.txt")
document = document_path.read_text(encoding="utf-8")
```

```python
model = "gemma3"
```

```python
prompt = f"""
You are a document analysis assistant.

Answer questions using only the document below.
If the document does not contain enough information to answer a question,
say so rather than making up an answer.

DOCUMENT:
{document}
"""

messages = [
    {"role": "system", "content": prompt}
]
```

```python
def ask(question):
    messages.append({"role": "user", "content": question})

    response = ollama.chat(
        model=model,
        messages=messages
    )

    answer = response["message"]["content"]
    messages.append({"role": "assistant", "content": answer})

    print(f"You: {question}\n")
    print(f"Gemma 3: {answer}")
    return answer
```
